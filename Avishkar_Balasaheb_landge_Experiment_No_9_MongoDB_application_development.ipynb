## Aim

To develop a Python-based Hospital Management System using MongoDB as the backend database and implement CRUD operations—Create, Read, Update, and Delete—for hospital entities such as departments, doctors, patients, appointments, and medical reports using Google Colab and the PyMongo library.

## Theory

MongoDB is a NoSQL, document-oriented database that stores data in JSON-like documents. It provides flexible storage and is suitable for applications where data can have different structures.

PyMongo is the Python driver used to connect Python applications with MongoDB. It provides functions to perform CRUD operations such as inserting, reading, updating, and deleting documents.

In this practical, a Hospital Management System was developed using MongoDB. The database contains collections for departments, doctors, patients, appointments, and medical reports. The PyMongo library was used to insert and retrieve records, update existing information, and delete records. Thus, the practical demonstrates how MongoDB can be used as a backend database for a Python-based hospital application.

```python
# READ: Find one patient

patient = patients.find_one({"patient_id": "P001"})

print("Patient found:")
print(patient)
```

```python
# UPDATE: Modify patient information

result = patients.update_one(
    {"patient_id": "P001"},
    {"$set": {"city": "Mumbai"}}
)

print("Documents modified:", result.modified_count)
```

```python
patient = patients.find_one({"patient_id": "P001"})
print(patient)
```

```python
# UPDATE MANY: Increase experience of all doctors

result = doctors.update_many(
    {},
    {"$inc": {"experience": 1}}
)

print("Doctors updated:", result.modified_count)
```

```python
# Insert a temporary patient for demonstrating DELETE

patients.insert_one({
    "patient_id": "P005",
    "name": "Test Patient",
    "age": 30,
    "gender": "Male",
    "city": "Pune"
})

print("Temporary patient inserted.")
```

```python
# DELETE: Remove the temporary patient

result = patients.delete_one(
    {"patient_id": "P005"}
)

print("Documents deleted:", result.deleted_count)
```

```python
print(patients.find_one({"patient_id": "P005"}))
```
