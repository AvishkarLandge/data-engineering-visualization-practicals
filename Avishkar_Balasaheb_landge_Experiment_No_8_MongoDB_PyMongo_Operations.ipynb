## Aim

## Theory

MongoDB is a NoSQL, document-oriented database that stores data in the form of JSON-like documents. Unlike relational databases, MongoDB stores related data in collections instead of tables.

PyMongo is the official Python driver used to connect Python applications with MongoDB. It provides functions for performing database operations such as inserting, searching, updating, replacing, deleting, sorting, counting, and aggregating documents.

In this practical, a MongoDB database named `college_db` and a collection named `students` are used. Different PyMongo functions are performed to demonstrate CRUD operations and data analysis. The aggregation operation is used to group students based on their branch and calculate the average marks and number of students in each branch.

```python
for student in students.find():
    print(student)
```

```python
student = students.find_one({"student_id": 103})

print(student)
```

```python
pune_students = students.find({"city": "Pune"})

for student in pune_students:
    print(student)
```

```python
high_scorers = students.find({"marks": {"$gt": 80}})

for student in high_scorers:
    print(student)
```

```python
students_sorted = students.find().sort("marks", 1)

for student in students_sorted:
    print(student["name"], "-", student["marks"])
```

```python
students_sorted = students.find().sort("marks", -1)

for student in students_sorted:
    print(student["name"], "-", student["marks"])
```

```python
# Delete the current collection
students.drop()

# Create the collection again
students = db["students"]

# Insert the original 7 documents only once
result = students.insert_many(students_data)

print("Collection reset successfully!")
print("Documents inserted:", len(result.inserted_ids))
```

```python
print("Total documents:", students.count_documents({}))
```

```python
result = students.update_one(
    {"student_id": 101},
    {"$set": {"marks": 90}}
)

print("Documents modified:", result.modified_count)
```

```python
student = students.find_one({"student_id": 101})
print(student)
```

```python
result = students.update_many(
    {"branch": "CSE"},
    {"$inc": {"marks": 2}}
)

print("Documents modified:", result.modified_count)
```

```python
result = students.replace_one(
    {"student_id": 102},
    {
        "student_id": 102,
        "name": "Ayush",
        "branch": "CSE",
        "semester": 4,
        "marks": 82,
        "city": "Mumbai"
    }
)

print("Documents replaced:", result.modified_count)
```

```python
print(students.find_one({"student_id": 102}))
```

```python
result = students.delete_one({"student_id": 107})

print("Documents deleted:", result.deleted_count)
```

```python
student = students.find_one({"student_id": 107})

print(student)
```

```python
print("Total documents:", students.count_documents({}))
```

```python
pipeline = [
    {
        "$group": {
            "_id": "$branch",
            "average_marks": {"$avg": "$marks"},
            "student_count": {"$sum": 1}
        }
    },
    {
        "$sort": {
            "average_marks": -1
        }
    }
]

results = students.aggregate(pipeline)

for result in results:
    print(result)
```

```python
print("Final number of documents:", students.count_documents({}))

for student in students.find().sort("student_id", 1):
    print(
        student["student_id"],
        student["name"],
        student["branch"],
        student["marks"]
    )
```
