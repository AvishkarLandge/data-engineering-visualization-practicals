## Aim

To fetch data from a public REST API using Python, parse the JSON response, transform the retrieved data into a structured tabular format, perform basic data validation and preprocessing, and store the transformed data for further analysis.


## Theory

A **REST API (Representational State Transfer Application Programming Interface)** allows applications to communicate and exchange data over the internet. APIs commonly return data in **JSON (JavaScript Object Notation)** format, which is easy for programs to process.

In this experiment, Python was used to fetch data from a public REST API using the `requests` library. The JSON response was then parsed and converted into a Pandas DataFrame using `pandas.json_normalize()`.

Basic data validation was performed by checking:

* Missing values
* Duplicate records
* Data types
* Number of rows and columns

The data was then preprocessed by standardizing column names and removing duplicate records. Finally, the cleaned and transformed dataset was stored as a **CSV file** using Pandas' `to_csv()` function.

This process demonstrates a basic data pipeline for collecting, preparing, and storing data before further analysis.

```python
import requests
import pandas as pd

url = "https://jsonplaceholder.typicode.com/users"

response = requests.get(url)

print("Status Code:", response.status_code)

data = response.json()

print(data[:2])
```

```python
df = pd.json_normalize(data)

print(df.head())
print("\nShape:", df.shape)
```

```python
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nData types:")
print(df.dtypes)

print("\nNumber of rows and columns:")
print(df.shape)
```

```python
df.columns = df.columns.str.replace(".", "_", regex=False)

df = df.drop_duplicates()

df = df.drop(columns=["address_geo_lat", "address_geo_lng"], errors="ignore")

print(df.head())
```

```python
df["valid_email"] = df["email"].str.contains("@", na=False)

print(df[["email", "valid_email"]])
```

```python
df.to_csv("users_transformed.csv", index=False)

print("Data successfully saved as users_transformed.csv")
```

```python
# Convert JSON data into a Pandas DataFrame
df = pd.json_normalize(data)

print("Data in tabular format:")
display(df)
```

```python
# Basic data validation

print("Number of rows and columns:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nData types:")
print(df.dtypes)
```

```python
# Preprocessing

# Make column names simple
df.columns = df.columns.str.replace(".", "_", regex=False)

# Remove duplicate records
df = df.drop_duplicates()

print("Preprocessed Data:")
display(df)
```

```python
# Save the transformed data as CSV

df.to_csv("users_transformed.csv", index=False)

print("Data successfully saved as users_transformed.csv")
```
